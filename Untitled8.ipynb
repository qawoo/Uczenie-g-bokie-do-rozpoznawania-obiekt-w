```python
import csv
import json
import shutil
from datetime import date
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import torch
import torch.nn as nn
from sklearn.metrics import confusion_matrix
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

SEED = 0
EPOCHS = 8
LR = 1e-3
BATCH_SIZE = 128
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
OUTPUT_DIR = Path("wyniki_lab1")
OUTPUT_DIR.mkdir(exist_ok=True)

np.random.seed(SEED)
torch.manual_seed(SEED)
print(f"PyTorch: {torch.__version__}; urządzenie: {DEVICE}")
```

```python
MEAN = 0.2860
STD = 0.3530
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((MEAN,), (STD,)),
])

train_ds = datasets.FashionMNIST("./data", train=True, download=True, transform=transform)
test_ds = datasets.FashionMNIST("./data", train=False, download=True, transform=transform)
classes = train_ds.classes

test_dl = DataLoader(
    test_ds,
    batch_size=256,
    shuffle=False,
    pin_memory=DEVICE.type == "cuda",
)

print("Klasy:", classes)
print(f"Trening: {len(train_ds)}, test: {len(test_ds)}")
```

```python
ARCHITECTURES = {
    "A — mały": (128,),
    "B — bazowy": (256, 128),
    "C — duży": (512, 256, 128),
}


class MLP(nn.Module):
    def __init__(self, in_dim=28 * 28, hidden=(256, 128), n_classes=10):
        super().__init__()
        layers = [nn.Flatten()]
        current_dim = in_dim
        for hidden_dim in hidden:
            layers.extend([nn.Linear(current_dim, hidden_dim), nn.ReLU()])
            current_dim = hidden_dim
        layers.append(nn.Linear(current_dim, n_classes))
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x)


def count_parameters(model):
    return sum(param.numel() for param in model.parameters() if param.requires_grad)


for name, hidden in ARCHITECTURES.items():
    network = MLP(hidden=hidden)
    print(f"{name:12s} {str(hidden):18s} {count_parameters(network):>8,d} parametrów")
```

```python
@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    criterion = nn.CrossEntropyLoss(reduction="sum")
    loss_sum = 0.0
    correct = 0
    total = 0
    for images, labels in loader:
        images = images.to(DEVICE)
        labels = labels.to(DEVICE)
        logits = model(images)
        loss_sum += criterion(logits, labels).item()
        correct += (logits.argmax(dim=1) == labels).sum().item()
        total += labels.numel()
    return loss_sum / total, correct / total


def train_model(hidden):
    torch.manual_seed(SEED)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)

    generator = torch.Generator().manual_seed(SEED)
    train_dl = DataLoader(
        train_ds,
        batch_size=BATCH_SIZE,
        shuffle=True,
        generator=generator,
        pin_memory=DEVICE.type == "cuda",
    )
    model = MLP(hidden=hidden).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR)
    criterion = nn.CrossEntropyLoss()
    history = {"train_loss": [], "test_loss": [], "test_acc": []}

    for epoch in range(1, EPOCHS + 1):
        model.train()
        loss_sum = 0.0
        for images, labels in train_dl:
            images = images.to(DEVICE)
            labels = labels.to(DEVICE)
            optimizer.zero_grad()
            loss = criterion(model(images), labels)
            loss.backward()
            optimizer.step()
            loss_sum += loss.item() * labels.numel()

        train_loss = loss_sum / len(train_ds)
        test_loss, test_accuracy = evaluate(model, test_dl)
        history["train_loss"].append(train_loss)
        history["test_loss"].append(test_loss)
        history["test_acc"].append(test_accuracy)
        print(
            f"Epoka {epoch:2d}/{EPOCHS} | train_loss: {train_loss:.4f} "
            f"| test_loss: {test_loss:.4f} | accuracy: {100 * test_accuracy:.2f}%"
        )

    return model, history
```

```python
results = []
histories = {}
models = {}

for name, hidden in ARCHITECTURES.items():
    print(f"\nTrenowanie: {name} {hidden}")
    model, history = train_model(hidden)
    models[name] = model
    histories[name] = history
    results.append({
        "nazwa": name,
        "architektura": str(hidden),
        "parametry": count_parameters(model),
        "accuracy": history["test_acc"][-1],
        "test_loss": history["test_loss"][-1],
    })

best_result = max(results, key=lambda row: row["accuracy"])
best_name = best_result["nazwa"]
best_model = models[best_name]
best_history = histories[best_name]

print("\nPODSUMOWANIE:")
print(f"{'Wariant':14s} {'Architektura':19s} {'Parametry':>10s} {'Accuracy':>10s}")
for row in results:
    print(
        f"{row['nazwa']:14s} {row['architektura']:19s} "
        f"{row['parametry']:10,d} {100 * row['accuracy']:9.2f}%"
    )
print(f"Najwyższe accuracy: {best_name}, {100 * best_result['accuracy']:.2f}%")
```

```python
epochs = range(1, EPOCHS + 1)
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(epochs, best_history["train_loss"], "o-", label="Trening")
ax.plot(epochs, best_history["test_loss"], "o-", label="Test")
ax.set(title=f"Krzywe uczenia — {best_name}", xlabel="Epoka", ylabel="Strata")
ax.set_xticks(list(epochs))
ax.grid(alpha=0.25)
ax.legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "krzywe_uczenia.png", dpi=170)
plt.show()
```

```python
@torch.no_grad()
def get_predictions(model, loader):
    model.eval()
    images_all = []
    labels_all = []
    predictions_all = []
    for images, labels in loader:
        prediction = model(images.to(DEVICE)).argmax(dim=1).cpu()
        images_all.append(images.cpu())
        labels_all.append(labels.cpu())
        predictions_all.append(prediction)
    return torch.cat(images_all), torch.cat(labels_all), torch.cat(predictions_all)


images, labels, predictions = get_predictions(best_model, test_dl)
correct_ids = torch.where(predictions == labels)[0][:6].tolist()
wrong_ids = torch.where(predictions != labels)[0][:6].tolist()
example_ids = correct_ids + wrong_ids

fig, axes = plt.subplots(2, 6, figsize=(15, 6))
for ax, idx in zip(axes.flat, example_ids):
    image = (images[idx, 0] * STD + MEAN).clamp(0, 1)
    true_label = classes[labels[idx].item()]
    predicted_label = classes[predictions[idx].item()]
    correct = labels[idx] == predictions[idx]
    ax.imshow(image, cmap="gray", vmin=0, vmax=1)
    ax.set_title(
        f"Pred.: {predicted_label}\nPrawda: {true_label}",
        fontsize=8,
        color="green" if correct else "red",
    )
    ax.axis("off")
fig.suptitle("Górny rząd: trafne predykcje | Dolny rząd: błędne predykcje")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "przykladowe_predykcje.png", dpi=170)
plt.show()
```

```python
cm = confusion_matrix(labels.numpy(), predictions.numpy(), labels=list(range(len(classes))))
fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(cm, cmap="Blues")
ax.set_xticks(np.arange(len(classes)), labels=classes, rotation=50, ha="right")
ax.set_yticks(np.arange(len(classes)), labels=classes)
ax.set_xlabel("Przewidziana klasa")
ax.set_ylabel("Prawdziwa klasa")
ax.set_title(f"Macierz pomyłek — {best_name}")
fig.colorbar(im, ax=ax, label="Liczba obrazów")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "macierz_pomylek.png", dpi=170)
plt.show()

confusions = [
    (int(cm[i, j]), i, j)
    for i in range(len(classes))
    for j in range(len(classes))
    if i != j
]
confusions.sort(reverse=True)
print("Najczęstsze pomyłki (prawdziwa klasa -> przewidziana):")
for count, i, j in confusions[:5]:
    print(f"{classes[i]} -> {classes[j]}: {count} obrazów")

pairs = [
    (int(cm[i, j] + cm[j, i]), i, j)
    for i in range(len(classes))
    for j in range(i + 1, len(classes))
]
pair_count, pair_a, pair_b = max(pairs)
print(
    f"Najtrudniejsza para klas: {classes[pair_a]} / {classes[pair_b]} "
    f"({pair_count} pomyłek w obu kierunkach)"
)
```
